## Abrir y cerrar archivos

Para leer o escribir un archivo hay que **abrirlo** primero con `open()` y **cerrarlo** cuando terminamos:

```python
archivo = open('datos.txt', 'r')   # abrir para lectura
contenido = archivo.read()
archivo.close()                    # siempre cerrar
```

El segundo argumento es el **modo**:

| Modo | Descripción |
|------|-------------|
| `'r'` | Lectura (default). Error si no existe. |
| `'w'` | Escritura. Crea el archivo o **borra el contenido** si ya existe. |
| `'a'` | Agregar (*append*). Crea el archivo o escribe al final si ya existe. |
| `'x'` | Creación exclusiva. Error si ya existe. |

**Problema:** si ocurre un error entre `open()` y `close()`, el archivo queda abierto. Por eso usamos `with`.

---
## `with`: la forma correcta

El bloque `with` se encarga de cerrar el archivo automáticamente, incluso si hay un error:

```python
with open('datos.txt', 'r') as archivo:
    contenido = archivo.read()
# acá el archivo ya está cerrado
```

**Regla:** usá siempre `with` para abrir archivos. No hay razón para no hacerlo.

---
## Métodos de lectura

Hay tres formas principales de leer un archivo:

```python
# 1) read() — todo el contenido como un solo string
with open('datos.txt') as f:
    todo = f.read()

# 2) readlines() — lista de líneas (cada una con '\n' al final)
with open('datos.txt') as f:
    lineas = f.readlines()

# 3) Iterar línea por línea (la más eficiente para archivos grandes)
with open('datos.txt') as f:
    for linea in f:
        print(linea.strip())   # strip() saca el '\n'
```

También existe `readline()` que lee *una sola línea* cada vez que se llama.

---
## `strip()` y sus variantes

Al leer un archivo, cada línea viene con un `'\n'` al final. El método `strip()` elimina espacios en blanco (espacios, `\n`, `\t`) de ambos extremos:

```python
'  hola  '.strip()      # 'hola'
'hola\n'.strip()         # 'hola'
'\t hola \n'.strip()     # 'hola'
```

Variantes:

```python
'  hola  '.lstrip()     # 'hola  '   (solo izquierda)
'  hola  '.rstrip()     # '  hola'   (solo derecha)
```

Si solo querés sacar el salto de línea sin tocar otros espacios, usá `rstrip('\n')` en vez de `strip()`.

También acepta un argumento con los caracteres a eliminar:

```python
'***hola***'.strip('*')       # 'hola'
'abcholacba'.strip('abc')     # 'hola'
```

**Atención:** el argumento no es un substring sino un *conjunto de caracteres*. En el último ejemplo se eliminan `'a'`, `'b'` y `'c'` individualmente, no la secuencia `"abc"`.

---
## Escritura

Para escribir usamos `write()` o `print()` con el argumento `file`:

```python
# Con write() — no agrega '\n' automáticamente
with open('salida.txt', 'w') as f:
    f.write('primera línea\n')
    f.write('segunda línea\n')

# Con print() — sí agrega '\n'
with open('salida.txt', 'w') as f:
    print('primera línea', file=f)
    print('segunda línea', file=f)
```

**Cuidado:** el modo `'w'` borra todo el contenido anterior. Si querés agregar al final, usá `'a'`.

## Archivos CSV manuales

Un archivo CSV (*Comma-Separated Values*) guarda datos tabulares como texto, separando valores con comas. La primera línea suele ser el encabezado.

```
nombre,edad,ciudad
Ana,30,Córdoba
Bruno,25,Rosario
```

Para procesarlo manualmente, leemos línea por línea y usamos `split(',')`:

```python
with open('datos.csv') as f:
    encabezado = f.readline().strip().split(',')
    for linea in f:
        campos = linea.strip().split(',')
        # hacer algo con campos
```

Python tiene un módulo `csv` para casos más complejos (valores con comas adentro, comillas, etc.), pero para archivos simples alcanza con `split()`.

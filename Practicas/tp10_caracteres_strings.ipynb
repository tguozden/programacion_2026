## chr() y ord()

Cada carácter tiene un número entero asociado según el estándar Unicode. `ord()` y `chr()` convierten entre los dos:

```python
ord('A')   # 65
ord('a')   # 97
ord('0')   # 48

chr(65)    # 'A'
chr(97)    # 'a'
chr(9829)  # '♥'
```

Los rangos más útiles:

| Rango | Caracteres |
|-------|------------|
| 48–57 | `'0'`–`'9'` |
| 65–90 | `'A'`–`'Z'` |
| 97–122 | `'a'`–`'z'` |

Las letras mayúsculas y minúsculas difieren exactamente en 32:

```python
ord('a') - ord('A')   # 32
chr(ord('A') + 32)    # 'a'
```

## Strings como secuencias

Un string es una secuencia inmutable de caracteres. Podés recorrerlo, indexarlo y slicearlo igual que una lista, pero no modificarlo.

```python
s = 'hola'
s[0]       # 'h'
s[-1]      # 'a'
s[1:3]     # 'ol'
s[::-1]    # 'aloh'

for c in s:
    print(c)
```

Para construir un string nuevo desde cero, la forma más limpia es acumular en una lista y unir al final:

```python
resultado = []
for c in s:
    resultado.append(c.upper())
''.join(resultado)   # 'HOLA'
```

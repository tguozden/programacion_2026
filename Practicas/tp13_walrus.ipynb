## El operador `:=`

El **operador morsa** (`:=`, *walrus operator*) permite asignar un valor a una variable *dentro* de una expresión. Se llama así porque `:=` se parece a los colmillos de una morsa 🦭.

El problema que resuelve es el patrón "pedir, verificar, usar" que obliga a repetir código:

```python
# Sin morsa: el input aparece dos veces
numero = int(input('Ingresá un número: '))
while numero != 0:
    print(numero)
    numero = int(input('Ingresá un número: '))

# Con morsa: el input aparece una sola vez
while (numero := int(input('Ingresá un número: '))) != 0:
    print(numero)
```

La asignación con `:=` devuelve el valor asignado, por eso puede usarse dentro de un `while` o un `if`.

**Atención:** los paréntesis alrededor de la asignación son obligatorios cuando se usa dentro de un `while` o `if`.

## 1
### ¿Qué imprime?

Antes de ejecutar, escribí tu respuesta como comentario.

````python
# a)
x = 0
if (x := 5 + 3) > 6:
    print(x)

# b)
numeros = [4, 1, 7, 2, 9, 3]
grandes = [y for n in numeros if (y := n * 2) > 10]
print(grandes)

# c)
datos = [0, 0, 3, 0, 5]
i = 0
while not (valor := datos[i]):
    i += 1
print(valor, i)
````

## 2
### Reescribir con morsa

Reescribí cada bloque usando el operador morsa para eliminar la repetición.

````python
# a) El input aparece dos veces
numero = int(input('Ingresá un número (0 para terminar): '))
total = 0
while numero != 0:
    total += numero
    numero = int(input('Ingresá un número (0 para terminar): '))
print('Total:', total)

# --- Tu versión con morsa acá ---


# b) El input aparece dos veces
palabra = input("Ingresá una palabra ('fin' para terminar): ")
largas = []
while palabra != 'fin':
    if len(palabra) > 4:
        largas.append(palabra)
    palabra = input("Ingresá una palabra ('fin' para terminar): ")
print('Palabras largas:', largas)

# --- Tu versión con morsa acá ---
````

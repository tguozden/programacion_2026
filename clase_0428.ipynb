```python
lista = [1, 2, 5, 6, 7, 8, 2, 5, 3, 3, 8, 4, 10, 40, -4, 5]
print(lista)
for j in range(len(lista)):
    
    sublista = lista[j:]
    iminimo = 0
    minimo = sublista[0]
    for i in range(len(sublista)):
        if sublista[i]<minimo:
            minimo = sublista[i]
            iminimo = i
    # swap(0, iminimo, lista) # ejercicio!
    aux = sublista[0]
    sublista[0] = sublista[iminimo]
    sublista[iminimo] = aux
    lista[j:] = sublista
    print(lista)
    
```

## Errores y excepciones

Cuando Python encuentra un error durante la ejecución, lanza una **excepción** y el programa se detiene. Cada tipo de error tiene su propio nombre:

```python
int('hola')          # ValueError
1 / 0                # ZeroDivisionError
[1, 2, 3][10]        # IndexError
{'a': 1}['z']        # KeyError
open('noexiste.txt') # FileNotFoundError
int(None)            # TypeError
```

El bloque `try`/`except` permite **atrapar** la excepción y decidir qué hacer en lugar de que el programa se caiga.

---
## `try` / `except`

La estructura básica:

```python
try:
    # código que puede fallar
    numero = int(input('Ingresá un número: '))
    print('El doble es', numero * 2)
except ValueError:
    print('Eso no es un número')
```

Si el código dentro del `try` lanza un `ValueError`, la ejecución salta al `except`. Si no hay error, el `except` se ignora.

Se pueden atrapar varios tipos de excepción:

```python
try:
    resultado = lista[i] / divisor
except IndexError:
    print('Índice fuera de rango')
except ZeroDivisionError:
    print('No se puede dividir por cero')
```

También se puede agrupar:

```python
except (ValueError, TypeError):
    print('Dato inválido')
```

Para acceder al mensaje de error, usá `as` para guardarlo en una variable:

```python
try:
    int('hola')
except ValueError as e:
    print('Falló:', e)
# Falló: invalid literal for int() with base 10: 'hola'
```

Si querés atrapar cualquier error sin importar el tipo, usá `Exception`:

```python
try:
    # algo que puede fallar de muchas formas
except Exception as e:
    print(f'Error inesperado: {e}')
```

**Regla:** siempre que puedas, atrapá excepciones específicas. `except Exception` es útil como red de seguridad (por ejemplo, dentro de un bucle que no debe detenerse), pero si lo usás en todos lados esconde bugs.

---
## `else` y `finally`

El bloque completo tiene cuatro partes:

```python
try:
    # código que puede fallar
except TipoDeError:
    # qué hacer si falla
else:
    # se ejecuta solo si NO hubo error
finally:
    # se ejecuta SIEMPRE, haya o no error
```

Ejemplo:

```python
try:
    archivo = open('datos.txt')
except FileNotFoundError:
    print('El archivo no existe')
else:
    contenido = archivo.read()
    archivo.close()
    print('Leído OK')
finally:
    print('Fin del intento')
```

`else` y `finally` son opcionales. `else` es útil para separar el código que puede fallar del que no. `finally` sirve para limpieza (cerrar archivos, conexiones, etc.).

## `raise`: lanzar excepciones

Podés lanzar una excepción vos mismo con `raise`:

```python
def dividir(a, b):
    if b == 0:
        raise ValueError('El divisor no puede ser cero')
    return a / b
```

Esto es útil para validar datos en tus propias funciones. Quien llame a la función puede atrapar la excepción con `try`/`except`.

```python
import random
aleat = []
for i in range(10):
    aleat.append(random.randint(1,10))
```

```python
print(aleat)
imin = 0
vmin = aleat[imin]
for i in range(len(aleat)):
    if aleat[i]< vmin:
        imin = i
        vmin = aleat[i]
print(imin,vmin)
```

```python
def mihash(cadena:str)->int:
    acum = 1
    for i in cadena:
        acum*=ord(i)
    return acum%123
mihash('el lobo está en el bosque, cuando vos no estás')
```

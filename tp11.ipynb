## Argumentos posicionales y nombrados

Al llamar a una función, podés pasar los argumentos **por posición** o **por nombre**:

```python
def presentar(nombre, edad):
    print(f'{nombre} tiene {edad} años')

presentar('Ana', 30)           # posicional
presentar(edad=30, nombre='Ana')  # nombrado (el orden no importa)
presentar('Ana', edad=30)      # mixto: posicional primero, nombrado después
```

**Regla:** los posicionales siempre van antes que los nombrados.

---
## Valores por defecto

Un parámetro puede tener un valor por defecto. Si no se pasa ese argumento, se usa el default:

```python
def saludar(nombre, saludo='Hola'):
    print(f'{saludo}, {nombre}!')

saludar('Ana')              # 'Hola, Ana!'
saludar('Ana', 'Buenas')   # 'Buenas, Ana!'
saludar('Ana', saludo='Hey')  # 'Hey, Ana!'
```

**Regla:** los parámetros con default siempre van después de los sin default.

---
## `*args`: cantidad variable de argumentos posicionales

`*args` empaqueta todos los argumentos posicionales extra en una **tupla**:

```python
def sumar(*numeros):
    total = 0
    for n in numeros:
        total += n
    return total

sumar(1, 2, 3)      # 6
sumar(10, 20)       # 30
sumar()             # 0
```

También podés tener parámetros normales antes:

```python
def imprimir_lista(titulo, *items):
    print(titulo)
    for item in items:
        print(' -', item)

imprimir_lista('Frutas', 'manzana', 'pera', 'uva')
```

---
## `**kwargs`: cantidad variable de argumentos nombrados

`**kwargs` empaqueta todos los argumentos nombrados extra en un **diccionario**:

```python
def mostrar_info(**datos):
    for clave, valor in datos.items():
        print(f'{clave}: {valor}')

mostrar_info(nombre='Ana', edad=30, ciudad='Córdoba')
# nombre: Ana
# edad: 30
# ciudad: Córdoba
```

El orden de los parámetros en una firma completa es:
```python
def f(posicionales, *args, con_default=valor, **kwargs):
    ...
```

---
## `/` y `*`: forzar cómo se pasan los argumentos

Python permite restringir si un parámetro se puede pasar por posición, por nombre, o ambos:

```python
def f(a, b, /, c, d, *, e, f):
    ...
```

| Zona | Separador | Cómo se puede pasar |
|------|-----------|---------------------|
| antes de `/` | `/` | **solo posicional** |
| entre `/` y `*` | — | posicional o nombrado |
| después de `*` | `*` | **solo nombrado** |

```python
def f(a, /, b, *, c):
    print(a, b, c)

f(1, 2, c=3)     # OK
f(1, b=2, c=3)   # OK
f(a=1, b=2, c=3) # Error: a es solo posicional
f(1, 2, 3)       # Error: c es solo nombrado
```

Firma completa con todo:
```python
def f(solo_pos, /, normales, *args, solo_nom, **kwargs):
    ...
```

## 9
### Propagación con `*` y `**`

El operador `*` también sirve para **desempaquetar** una lista al llamar una función, y `**` para desempaquetar un diccionario.

```python
def f(a, b, c):
    return a + b + c

args = [1, 2, 3]
f(*args)          # equivale a f(1, 2, 3)

kwargs = {'a': 1, 'b': 2, 'c': 3}
f(**kwargs)       # equivale a f(a=1, b=2, c=3)
```

¿Qué imprime el siguiente código? Respondé antes de ejecutar.

````python
# Antes de ejecutar, escribí tu respuesta como comentario.

def resta(a, b, c):
    return a - b - c

valores = [10, 3, 2]
print(resta(*valores))

parametros = {'c': 2, 'a': 10, 'b': 3}
print(resta(**parametros))

print(resta(*valores[:2], **{'c': 2}))
````

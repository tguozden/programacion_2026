## Hasheables

Un objeto es **hasheable** si tiene un valor de hash fijo durante toda su vida. Python necesita esto para usarlo como clave de diccionario o elemento de conjunto.

```python
hash(42)        # funciona
hash('hola')    # funciona
hash((1, 2))    # funciona
hash([1, 2])    # TypeError: unhashable type: 'list'
```

La regla no es simplemente "las tuplas son hasheables". Es: **un objeto es hasheable si es inmutable *y* todo su contenido también lo es**.

```python
(1, 2, 3)    # hasheable: todos los elementos son inmutables
(1, [2, 3])  # NO hasheable: contiene una lista
```

¿Por qué importa? Porque el hash se usa para ubicar el objeto en memoria. Si el contenido pudiera cambiar, el hash cambiaría y el diccionario perdería la clave.

## Desempaquetado (*unpacking*)

**Desempaquetado** es asignar los elementos de una secuencia a variables individuales en una sola línea.

```python
punto = (3, 7)
x, y = punto          # x = 3, y = 7

a, b, c = [10, 20, 30]
```

El operador `*` aparece en dos contextos y hace cosas opuestas según de qué lado de la asignación esté:

**A la izquierda — captura:** agrupa el resto de los elementos en una lista.

```python
primero, *resto = [1, 2, 3, 4, 5]
# primero = 1
# resto   = [2, 3, 4, 5]   ← siempre es una lista

*inicio, ultimo = [1, 2, 3, 4, 5]
# inicio = [1, 2, 3, 4]
# ultimo = 5

primero, *medio, ultimo = [1, 2, 3, 4, 5]
# primero = 1
# medio   = [2, 3, 4]
# ultimo  = 5
```

Solo puede haber **un `*`** en cada asignación.

**A la derecha — expande:** desempaqueta una secuencia en elementos separados, como si los hubieras escrito uno por uno.

```python
numeros = [1, 2, 3]
print(*numeros)        # equivale a print(1, 2, 3)
```

El desempaquetado también funciona en bucles:

```python
pares = [(1, 'a'), (2, 'b'), (3, 'c')]
for numero, letra in pares:
    print(numero, letra)
```

Y permite hacer un swap sin variable auxiliar (Python primero evalúa el lado derecho como tupla, después desempaqueta):

```python
a, b = b, a
```

El `*` a la derecha solo funciona dentro de un contenedor o llamada a función — no puede quedar suelto:

```python
print(*'abc')        # funciona: a b c
[*'abc']             # funciona: ['a', 'b', 'c']
(*'abc',)            # funciona: ('a', 'b', 'c')
{*'abc'}             # funciona: {'a', 'b', 'c'}

a = *('abc')         # SyntaxError
```

Esto significa que `[*iterable]` es equivalente a `list(iterable)`, y permite combinar iterables en una línea:

```python
a = [1, 2, 3]
b = [4, 5, 6]
[*a, *b]             # [1, 2, 3, 4, 5, 6]
```

## enumerate

`enumerate(iterable)` devuelve una secuencia de tuplas `(índice, elemento)` para cada elemento del iterable.

```python
letras = ['a', 'b', 'c']

for par in enumerate(letras):
    print(par)
# (0, 'a')
# (1, 'b')
# (2, 'c')
```

Como cada elemento es una tupla, podemos desempaquetar directamente en el `for`:

```python
for i, letra in enumerate(letras):
    print(i, letra)
```

Podés cambiar el índice inicial con `start`:

```python
for i, letra in enumerate(letras, start=1):
    print(i, letra)   # 1 a, 2 b, 3 c
```

## zip

`zip(a, b)` combina dos iterables elemento a elemento y devuelve una secuencia de tuplas. Se detiene cuando el más corto termina.

```python
nombres = ['Ana', 'Bruno', 'Carla']
notas   = [8, 5, 9]

for par in zip(nombres, notas):
    print(par)
# ('Ana', 8)
# ('Bruno', 5)
# ('Carla', 9)
```

Como cada elemento es una tupla, podemos desempaquetar directamente en el `for`:

```python
for nombre, nota in zip(nombres, notas):
    print(nombre, nota)
```

`zip` también acepta más de dos iterables:

```python
list(zip([1,2,3], ['a','b','c'], [True, False, True]))
# [(1, 'a', True), (2, 'b', False), (3, 'c', True)]
```

El `*` a la derecha permite "descomprimir" una lista de tuplas en argumentos separados para `zip`:

```python
datos = [('Ana', 25), ('Bruno', 30), ('Carla', 35)]
nombres, edades = zip(*datos)   # * expande la lista en argumentos
print(nombres)  # ('Ana', 'Bruno', 'Carla')
print(edades)   # (25, 30, 35)
```

Es el mismo `*` de siempre: expande la secuencia como si hubieras escrito cada elemento por separado.

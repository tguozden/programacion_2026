```python
import math
# x=[i*.1 for i in range(100)] # x va de 0 a 9.9
x = []
for i in range(100):
    x.append(i/10)
    
ys=[math.sin(i) for i in x]
yc=[math.cos(i) for i in x]
yt=[math.tan(i) for i in x]

plt.figure(figsize=(4,2))
plt.plot(x,ys, label = 'sin(x)')
plt.plot(x,yc, label = 'cos(x)')
plt.plot(x,yt, label = 'tan(x)')
plt.ylim(top=4, bottom = -4)
plt.xlim(left=0, right=3.14)
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
```

```python
fig, axes = plt.subplots(1,2, figsize=(4,2))
ax1 = axes[0]
ax2 = axes[1]
ax1.plot(x,ys)
ax2.plot(x,yc)
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(4,2), sharex=True, sharey=True)
ax00 = axes[0][0]
ax01 = axes[0][1]
ax10 = axes[1][0]
ax11 = axes[1][1]
ax00.plot(x,ys)
ax01.plot(x,yc)
ax10.plot(x,yt)
ax11.plot(x,ys)
ax11.plot(x,yc)
ax11.plot(x,yt)
ax11.set_xlim(0,5)
ax11.set_ylim(-2,2)
plt.savefig('trigono.pdf')
```

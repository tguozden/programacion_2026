```python
lista_dicc_satelites = []
for i in satelites:
    diccio = {'nombre': i.nombre, 'alt': i.altitud_km, 'orb': i.tipo_orbita() }
    lista_dicc_satelites.append(diccio)
```

```python
lista_dicc_satelites[:3]
```

```python
df = pd.DataFrame(lista_dicc_satelites)
df.head()
```

```python
suma_leo = 0
cant_leo = 0
for i in lista_dicc_satelites:
    if i['orb'] == 'LEO':
        suma_leo += i['alt']
        cant_leo += 1
print('promedio = ', suma_leo/cant_leo)
```

```python
df.groupby(['orb'])['alt'].mean()
```

```python
df.loc[100]
```

```python
dfleo = df.loc[df['orb'] == 'LEO' ]
print(dfleo['alt'].mean(), dfleo['alt'].std())
```

```python
# otra manera de calcular .mean()
df.groupby(['orb'])['alt'].agg('mean')
```

```python
# para calcular más variables al agrupar:
# fíjense la estructura de los argumentos:
# 'mean', 'count', 'max' y 'min' son referencias a funciones propias de pandas
# el primer campo le da título al nuevo dataframe
df.groupby(['orb'])['alt'].agg(cantidad='count', promedio='mean', maximo='max',
                               minimo='min')
```

```python
# matriz: una lista de listas de mismo tamaño
matriz = [[1, 2, 3] for i in range(3)]
```

```python
print(matriz)
```

```python
n = len(matriz)  # cantidad de filas
m = len(matriz[0]) # cantidad de elementos de alguna fila
for i in range(n):
    for j in range(m):
        print(matriz[i][j], end=' ')
    print()
```

```python
def imprime_mat(matriz: list) -> None:
    '''imprime matrices
    '''
    n = len(matriz)  # cantidad de filas
    m = len(matriz[0]) # cantidad de elementos de alguna fila
    for i in range(n):
        for j in range(m):
            print(matriz[i][j], end=' ')
        print()
matriz = [[1+i, 2+i, 3+i] for i in range(3)]
imprime_mat(matriz)
```

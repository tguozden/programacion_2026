```python
#super()

class Observacion:
    def __init__(self, ciudad, temperatura):
        self.ciudad = ciudad
        self.temperatura = temperatura

    def temperatura_f(self):
        return self.temperatura * 9 / 5 + 32

    def __str__(self):
        return f"{self.ciudad}: {self.temperatura} °C"

    def __repr__(self):
        return f"Observacion({self.ciudad!r}, {self.temperatura!r})"


class otraObservacion(Observacion):
    def temperatura_f(self):
        tf = self.temperatura * 9 / 5 + 32
        print('antes', self.temperatura, 'despues', tf)
        return tf
```

```python
obj = otraObservacion("esquel", 10)
obj.temperatura_f() #me devuelva el numero y además imprima algo
```

```python
class superObservacion(Observacion):
    def temperatura_f(self):
        tf = self.temperatura * 9 / 5 + 32
        print('antes', self.temperatura, 'despues', tf)
        super().temperatura_f()  # dejo que el método viejo siga al volante

        
obj = otraObservacion("tecka", 20)
obj.temperatura_f() #me devuelva el numero y además imprima algo
```

```python
# Cree una matriz 8x8 y rellénala con 0s y 1s siguiendo 
# el patrón de blancas y negras de un tablero de ajedrez
a = np.zeros((8,8))
```

```python
a.shape
```

```python
for i in range(8):
    for j in range(8):
        if (i+j)%2:
            a[i,j]=1
a.astype('int')
```

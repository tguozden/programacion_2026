```python
def esprimo(n:int)->bool:
    primo = True
    for i in range(2,n):
        if n%i == 0:
            primo = False
    return primo
```

```python
lista_primos = []
for i in range(1,101):
    if esprimo(i):
        lista_primos += [i]
print(lista_primos)
```

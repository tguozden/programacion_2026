```python
import catalogo, satelite
import datetime
from matplotlib import pyplot as plt

ahora = datetime.datetime(2026,10,29,13)
registros = catalogo.leer_satelites('datos/satelites.json')
registros[0]['OBJECT_NAME']
```

```python
satelites = satelite.crear_satelites(registros,ahora)
```

```python
print(satelites[100])
```

```python
# cronología de lanzamientos
lanzamientos = catalogo.lanzamientos_por_anio(registros)
#lanzamientos
```

```python
fig = plt.figure(figsize = (5,3))
plt.bar(x=lanzamientos.keys(), height=lanzamientos.values())    # para graficar en escala logarítmita, log=True)
plt.show()
```

```python
reg_starlinks = [i for i in registros if catalogo.grupo(i['OBJECT_NAME'])=='Starlink']
lanz_starlinks = catalogo.lanzamientos_por_anio(reg_starlinks)
len(reg_starlinks)
```

```python
fig = plt.figure(figsize = (5,3))

plt.bar(x=lanzamientos.keys(), height=lanzamientos.values())
plt.bar(x=lanz_starlinks.keys(), height=lanz_starlinks.values())
```

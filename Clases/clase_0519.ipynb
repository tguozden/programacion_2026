Escribí una función recursiva sumar_digitos(n) que sume los dígitos de un número entero positivo, y seguí sumando hasta llegar a un solo dígito. No vale convertir a string.


```python
def sumar_digitos(n:int)->int:
    total = 0
    while n > 0:
        aux = int(n/10)
        cifra = n-aux*10
        # print(cifra)
        total += cifra
        n=aux
    return total
```

```python
sumar_digitos(123) # 1+2+3 = 6
```

```python
sumar_digitos(87387968923641)
```

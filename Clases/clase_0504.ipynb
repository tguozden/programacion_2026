```python
def miswap(lista, i, j):
    aux = lista[i]
    lista[i] = lista[j]
    lista[j] = aux
print(lista)
miswap(lista, 0, 1)
print(lista)
    
```

```python
lista = [2,7,4,-3,2,0,1]
# vamos a burbujear
print(lista)
for i in range(len(lista) - 1):
    print(f"i= {i}", end=' ')
    if lista[i] > lista[i+1]:
        miswap(lista, i, i + 1)
    print(lista)
    
```

```python
lista = [2,7,4,-3,2,0,1]
# vamos a burbujear
print(lista)
for j in range(len(lista) - 1):
    for i in range(len(lista) - 1 - j):
        print(f"j={j}, i={i}: ", end=' ')
        if lista[i] > lista[i+1]:
            miswap(lista, i, i + 1)
        print(lista)
```

```python
# funciones recursivas

# supongamos la siguiente función:
# suma(N) = N+ N-1 + N-2 + ... + 1
# cómo puedo implementarla?

def suma(n: int) -> int:
    acumulador = 0
    for i in range(n, 0, -1):
        acumulador += i
    return acumulador
suma(1)
```

```python
# observemos que
suma(n) = n + n - 1 + ... + 1
suma(n) = n + suma(n - 1)
suma(1) = 1

suma(4) = 4 + 3 + 2 + 1
suma(3) =     3 + 2 + 1
suma(4) = 4 + suma(3)
```

```python
def sumar(n:int)->int:
    if n > 1:
        return n + sumar(n-1)
    else:
        return 1
sumar(1)    
```

```python
sumar(4)
```

```python
# factorial recursivo
# n! = n*(n-1)*(n-2)*..*1
# 0! = 1
def factor(n:int)->int:
    if n <= 1:
        return 1
    else:
        return n * factor(n-1)
factor(4)        
        
```

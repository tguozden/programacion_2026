```python
texto = '''Un texto es un conjunto coherente de signos y enunciados, orales 
o escritos, que transmiten un mensaje con una intención comunicativa. 
Todo texto efectivo posee un sentido completo dentro de su contexto y 
se organiza respetando tres propiedades básicas: coherencia (unidad temática), 
cohesión (conexión entre sus partes) y adecuación (adaptación al destinatario).'''
lista = texto.split()
```

```python
# lo atamos con alambre, lo atamos
def filtra_chirimbolos(palabra:str)->str:
    aux=''
    for i in palabra:
        if i.isalpha():
            aux+=i
    return aux
lista_filtrada = [filtra_chirimbolos(palabra) for palabra in lista]    
```

```python
# un poquito más inteligente
def lista_de_texto(texto:str)->list:
    aux = ''.join([i for i in texto if i.isalpha() or i==' '])
    return aux.split()
print(lista_de_texto(texto))
```

```python
def frecuencias(lista:list)->dict:
    # lista1 = [len(i) for i in lista]
    # lista1.min()
    # lista1.max()
    frec = {i:0 for i in range(1,20)}
    for i in lista:
        frec[len(i)] +=1
    return frec
frecuencias(lista_de_texto(texto))
```

```python
def frecuencias2(lista:list)->dict:
    frec=dict()
    for i in lista:
        n = len(i)
        if n in frec.keys(): 
            frec[n] +=1
        else:
            frec[n] = 0  # nueva entrada en el diccionario           
    return frec
frecuencias2(lista_de_texto(texto))
```

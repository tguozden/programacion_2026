### Distancia al origen

La distancia euclidiana de un punto `(x, y)` al origen es:

$$d = \sqrt{x^2 + y^2}$$

Escribí una función `distancias(puntos)` que reciba la lista de tuplas y devuelva un diccionario donde cada clave es una tupla y su valor es la distancia al origen.

Resultado esperado:
```python
{(0.8, 0.2): 0.825..., (0.1, 0.5): 0.509..., ...}
```

La raíz cuadrada se puede calcular como `x ** 0.5`.

````python
def distancias(puntos:list)->dict:
    d = dict()
    for pto in puntos:
        dist = (pto[0]**2+pto[1]**2)**0.5
        d[pto] = dist
    return d
    # Tu código acá
````

## 3  (4 puntos)
### Centroide

El centroide de un conjunto de puntos es el punto promedio: se calcula como el promedio de todas las coordenadas `x` y el promedio de todas las coordenadas `y`.

Escribí una función `centroide(puntos)` que devuelva una tupla `(cx, cy)` con las coordenadas del centroide.

Resultado esperado para la cadena de arriba:
```python
(0.4, 0.433)
```

````python
puntos = [(0.8, 0.2), (0.1, 0.5), (0.6, 0.6), (0.2, 0.1), (0.4, 0.9), (0.3, 0.3)]
def centroide(puntos:list)->tuple:
    cx=0
    cy=0
    for pto in puntos:
        cx +=pto[0]
        cy +=pto[1]
    cx = cx /len(puntos)
    cy = cy /len(puntos)
    return cx,cy
centroide(puntos)    
````

````python
cx, cy = centroide(puntos)
print(f'Centroide: ({round(cx, 3)}, {round(cy, 3)})')  
````

### Orden por distancia

Escribí una función `orden_distancia(puntos)` que devuelva una lista de enteros indicando el orden de los puntos de menor a mayor distancia al origen.

Por ejemplo, si el punto más cercano al origen es el de la posición `3`, el primer elemento de la lista será `3`.

Resultado esperado para la cadena de arriba:
```python
[3, 5, 1, 0, 2, 4]
```

Pista: podés usar el diccionario `dists` del ejercicio anterior.

````python
puntos = [(0.8, 0.2), (0.1, 0.5), (0.6, 0.6), (0.2, 0.1), (0.4, 0.9), (0.3, 0.3)]

def orden_distancia(puntos:list)->list:
    lista = []
    for i in puntos:
        lista.append((i[0]**2+i[1]**2)**.5)
    return lista
orden_distancia(puntos)
````

````python
ds = [0.8246211251235323,
 0.5099019513592785,
 0.848528137423857,
 0.223606797749979,
 0.9848857801796105,
 0.4242640687119285]
def eje4(puntos:list)->list:
    ds = [(i[0]**2 +i[1]**2)**0.5 for i in puntos]
    ds_ord = ds.copy()
    ds_ord.sort()  # inplace!! modifica la lista
    # print(ds)
    # print(dscopia)
    return [ds.index(i) for i in ds_ord]
eje4(puntos)
````

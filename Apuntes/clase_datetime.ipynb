# Clase: manejo de fechas con `datetime`

En Python, el módulo `datetime` de la biblioteca estándar permite trabajar con fechas, horas, y duraciones (diferencias entre fechas).

Los objetos principales que vamos a usar son:

- `date`: representa una fecha (año, mes, día)
- `time`: representa una hora (hora, minuto, segundo, microsegundo)
- `datetime`: combina fecha y hora
- `timedelta`: representa una duración (diferencia entre dos fechas/horas)

Empecemos importando lo que vamos a necesitar.

```
from datetime import date, time, datetime, timedelta
```

## 1. Crear fechas y horas

Podemos crear un objeto `datetime` indicando año, mes, día, hora, minuto, segundo:

```
fecha = datetime(2026, 8, 13, 15, 30, 0)
print(fecha)
print(type(fecha))
```

También podemos crear solo una fecha (sin hora) con `date`:

```
d = date(2026, 8, 13)
print(d)
```

### Fecha y hora actual

- `datetime.now()` devuelve la fecha y hora actual
- `date.today()` devuelve solo la fecha actual (sin hora)

```
ahora = datetime.now()
print("Ahora:", ahora)

hoy = date.today()
print("Hoy:", hoy)
```

## 2. Atributos de un `datetime`

Un objeto `datetime` (o `date`) tiene atributos para acceder a cada componente:

```
fecha = datetime(2026, 8, 13, 15, 30, 45)

print("Año:", fecha.year)
print("Mes:", fecha.month)
print("Día:", fecha.day)
print("Hora:", fecha.hour)
print("Minuto:", fecha.minute)
print("Segundo:", fecha.second)
```

### Día de la semana

- `weekday()`: devuelve un número de 0 (lunes) a 6 (domingo)
- `isoweekday()`: devuelve un número de 1 (lunes) a 7 (domingo)

No hay un método que devuelva directamente el *nombre* del día en español; para eso usamos `strftime` (lo vemos a continuación) o armamos nosotros una lista de nombres.

```
fecha = datetime(2026, 8, 13)  # jueves

print("weekday():", fecha.weekday())        # 0=lunes ... 6=domingo
print("isoweekday():", fecha.isoweekday())  # 1=lunes ... 7=domingo

dias = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]
print("Día de la semana:", dias[fecha.weekday()])
```

## 3. Formatear fechas: `strftime`

`strftime` ("string format time") convierte un objeto `datetime` en una cadena de texto con el formato que le indiquemos, usando códigos especiales:

| Código | Significado              | Ejemplo |
|--------|---------------------------|---------|
| `%d`   | Día del mes (2 dígitos)   | 13      |
| `%m`   | Mes (2 dígitos)            | 08      |
| `%Y`   | Año (4 dígitos)            | 2026    |
| `%y`   | Año (2 dígitos)            | 26      |
| `%H`   | Hora (24hs)                | 15      |
| `%M`   | Minutos                    | 30      |
| `%S`   | Segundos                   | 45      |
| `%A`   | Nombre del día (en inglés) | Thursday|
| `%B`   | Nombre del mes (en inglés) | August  |

Por defecto, los nombres de días y meses aparecen en **inglés**, porque dependen de la configuración regional (*locale*) del sistema.

```
fecha = datetime(2026, 8, 13, 15, 30, 45)

print(fecha.strftime("%d/%m/%Y"))
print(fecha.strftime("%H:%M:%S"))
print(fecha.strftime("%A %d de %B del %Y"))
```

### Nombres de días y meses en español

Para obtener nombres en español sin depender del *locale* del sistema (que puede fallar o variar según la computadora), la forma más simple y portable es armar nosotros mismos las listas de nombres:

```
meses = ["enero", "febrero", "marzo", "abril", "mayo", "junio",
         "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"]
dias = ["lunes", "martes", "miércoles", "jueves", "viernes", "sábado", "domingo"]

fecha = datetime(2026, 8, 13)
nombre_dia = dias[fecha.weekday()]
nombre_mes = meses[fecha.month - 1]

print(f"{nombre_dia.capitalize()} {fecha.day} de {nombre_mes} del {fecha.year}")
```

## 4. Convertir texto a fecha: `strptime`

`strptime` ("string parse time") es la operación inversa: convierte una cadena de texto en un objeto `datetime`, indicando cuál es el formato del texto.

```
texto = "13/08/2026 15:30"
fecha = datetime.strptime(texto, "%d/%m/%Y %H:%M")
print(fecha)
print(type(fecha))
```

Esto es muy útil, por ejemplo, cuando leemos fechas desde un archivo CSV, donde vienen como texto y hay que convertirlas a `datetime` para poder operar con ellas.

## 5. Duraciones y aritmética de fechas: `timedelta`

Un `timedelta` representa una **diferencia** de tiempo (días, horas, minutos, etc.). Se puede sumar o restar a un `datetime` para obtener otra fecha, y también se obtiene al restar dos fechas entre sí.

```
hoy = date.today()

en_una_semana = hoy + timedelta(days=7)
print("Hoy:", hoy)
print("En una semana:", en_una_semana)

hace_10_dias = hoy - timedelta(days=10)
print("Hace 10 días:", hace_10_dias)
```

```
nacimiento = date(1998, 3, 20)
hoy = date.today()

diferencia = hoy - nacimiento
print(diferencia)
print("Días:", diferencia.days)
```

`timedelta` también permite sumar horas, minutos, semanas, etc.:

```
ahora = datetime.now()
en_90_minutos = ahora + timedelta(minutes=90)
print("Ahora:", ahora)
print("En 90 minutos:", en_90_minutos)
```

## 6. *Timestamp* y el *epoch*

El **epoch** (época) informático es el punto de referencia `1 de enero de 1970, 00:00:00 UTC`. Un *timestamp* es la cantidad de segundos transcurridos desde ese instante.

- `fecha.timestamp()`: convierte un `datetime` a su timestamp (float, segundos desde el epoch)
- `datetime.fromtimestamp(ts)`: convierte un timestamp a `datetime`

```
ahora = datetime.now()
ts = ahora.timestamp()
print("Timestamp actual:", ts)

de_vuelta = datetime.fromtimestamp(ts)
print("De vuelta a datetime:", de_vuelta)
```

**Un detalle importante:** si hacemos `datetime(1970,1,1).timestamp()`, el resultado *no* da exactamente 0. Esto pasa porque `datetime(1970,1,1)` se interpreta como una hora **local** (según la zona horaria de tu computadora), mientras que el epoch está definido en UTC. La diferencia que se ve es, justamente, el desfasaje horario (*offset*) de tu zona horaria respecto de UTC.

```
fecha = datetime(1970, 1, 1)
print(fecha.timestamp())
# Si tu zona horaria es UTC-3, vas a ver algo como 10800.0 (3 horas en segundos)
```

## 7. Comparar fechas

Los objetos `datetime` y `date` se pueden comparar directamente con `<`, `>`, `==`, igual que números:

```
fecha1 = date(2026, 8, 13)
fecha2 = date(2026, 12, 25)

print(fecha1 < fecha2)
print(fecha1 == date(2026, 8, 13))
```

## 8. Ejemplo integrador

Armemos una lista con los primeros lunes de cada mes del año 2026, en formato `"Lunes DD de mes del AAAA"`.

```
meses = ["enero", "febrero", "marzo", "abril", "mayo", "junio",
         "julio", "agosto", "septiembre", "octubre", "noviembre", "diciembre"]

primeros_lunes = []

for mes in range(1, 13):
    d = date(2026, mes, 1)
    # avanzamos hasta encontrar un lunes (weekday() == 0)
    while d.weekday() != 0:
        d += timedelta(days=1)
    primeros_lunes.append(f"Lunes {d.day:02d} de {meses[mes-1]} del {d.year}")

for l in primeros_lunes:
    print(l)
```

## Resumen

- `date`, `time`, `datetime`, `timedelta` son las herramientas básicas del módulo `datetime`.
- `strftime` convierte fecha → texto; `strptime` convierte texto → fecha.
- `weekday()` / `isoweekday()` dan el día de la semana como número.
- `timedelta` permite sumar/restar duraciones y también es el resultado de restar dos fechas.
- El *timestamp* cuenta segundos desde el epoch (1/1/1970 UTC); ojo con las zonas horarias.

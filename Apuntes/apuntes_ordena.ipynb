```python
# Paso 2: in-place (misma lógica, sin memoria extra)
def insertion_sort(lista):
    for j in range(1, len(lista)):
        clave = lista[j]
        i = j - 1
        while i >= 0 and lista[i] > clave:
            lista[i + 1] = lista[i]
            i -= 1
        lista[i + 1] = clave
        
```

```python
def bubble_sort(lista):
    n = len(lista)
    for pasada in range(n - 1):
        for i in range(n - 1 - pasada):
            if lista[i] > lista[i + 1]:
                swap(lista, i, i + 1)

def bubble_sort_flag(lista):
    n = len(lista)
    for pasada in range(n - 1):
        hubo_swap = False
        for i in range(n - 1 - pasada):
            if lista[i] > lista[i + 1]:
                swap(lista, i, i + 1)
                hubo_swap = True
        if not hubo_swap:
            break
        
```

```python
def selection_sort(lista):
    n = len(lista)
    for j in range(n - 1):
        imin = j
        for i in range(j + 1, n):
            if lista[i] < lista[imin]:
                imin = i
        if imin != j:
            swap(lista, j, imin)
```

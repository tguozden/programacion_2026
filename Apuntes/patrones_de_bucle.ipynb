# Patrones de bucle

Cuando resolvemos problemas con bucles, solemos usar los mismos **patrones** una y otra vez. Nombrarlos ayuda a reconocerlos rápido en problemas nuevos.

| Patrón | Para qué sirve |
|--------|---------------|
| Swap | intercambiar dos valores sin perder ninguno |
| Acumulador | sumar, contar, concatenar |
| Bandera | saber si algo ocurrió |
| Máximo / mínimo | encontrar el mejor valor (y su posición) |
| Variable auxiliar | comparar el elemento actual con el anterior |
| Secuencia más larga | rastrear una racha y su mejor versión |

---
## Patrón 0: Swap (intercambio)

**Problema:** queremos intercambiar los valores de dos variables. Si hacemos esto:

```python
a = 'rojo'
b = 'azul'
a = b    # ⚠️ perdimos 'rojo'!
b = a    # ahora b también es 'azul'
```

**Solución:** usar una variable auxiliar para no perder ningún valor:

```python
aux = a   # guardamos a antes de pisarla
a = b
b = aux   # recuperamos el valor original de a
```

En Python también existe una forma directa:

```python
a, b = b, a   # swap en una línea
```

El swap es el ejemplo más simple de **variable auxiliar**: una variable que existe solo para guardar un valor temporalmente mientras hacemos algo con él.

````python
# Swap con variable auxiliar
a = 'rojo'
b = 'azul'
print('Antes:', a, b)

aux = a
a = b
b = aux
print('Después:', a, b)  # azul rojo
````

## Patrón 1: Acumulador

**Idea:** empezamos con un valor neutro y lo vamos actualizando en cada iteración.

```python
total = 0          # valor inicial neutro para sumas
for x in lista:
    total += x     # actualización
```

El valor inicial depende de la operación:
- suma → `0`
- producto → `1`
- concatenar texto → `''`
- juntar lista → `[]`

````python
# Ejemplo: contar cuántos números son pares
numeros = [3, 8, 1, 4, 7, 2, 9, 6]

pares = 0
for n in numeros:
    if n % 2 == 0:
        pares += 1

print('Pares:', pares)  # 4
````

## Patrón 2: Bandera

**Idea:** una variable booleana que empieza en `False` y se vuelve `True` cuando ocurre algo. Sirve para responder preguntas de tipo *¿existe algún...?* o *¿ocurrió...?*

```python
encontrado = False          # la bandera empieza baja
for x in lista:
    if condicion(x):
        encontrado = True   # la bandera sube
        break               # (opcional) no hace falta seguir buscando

if encontrado:
    print('sí')
else:
    print('no')
```

````python
# Ejemplo: ¿hay algún número mayor a 100 en la lista?
numeros = [12, 45, 78, 23, 150, 9]

hay_grande = False
for n in numeros:
    if n > 100:
        hay_grande = True
        break

if hay_grande:
    print('Sí hay un número mayor a 100')
else:
    print('No hay ninguno')
````

## Patrón 3: Máximo / mínimo

**Idea:** guardamos el mejor valor visto hasta ahora. En cada paso, si el actual es mejor, lo reemplazamos.

```python
maximo = lista[0]    # arrancamos con el primero
for x in lista:
    if x > maximo:
        maximo = x
```

Si además necesitamos **la posición**, agregamos una variable más:

```python
maximo = lista[0]
pos_max = 0
for i in range(len(lista)):
    if lista[i] > maximo:
        maximo = lista[i]
        pos_max = i       # actualizamos las dos juntas
```

⚠️ Cuando actualizamos el máximo, **siempre** actualizamos su posición en el mismo `if`.

````python
# Ejemplo: temperatura más alta y en qué día
temps = [18, 22, 27, 25, 31, 29, 20]
dias  = ['lun', 'mar', 'mié', 'jue', 'vie', 'sáb', 'dom']

max_temp = temps[0]
dia_max  = 0
for i in range(len(temps)):
    if temps[i] > max_temp:
        max_temp = temps[i]
        dia_max  = i

print(f'Máxima: {max_temp}°C el {dias[dia_max]}')
````

## Patrón 4: Variable auxiliar (actual vs. anterior)

**Idea:** igual que en el swap, usamos una variable auxiliar — pero ahora dentro de un bucle. Guardamos el elemento anterior para poder compararlo con el actual en la próxima iteración.

```python
anterior = lista[0]              # la variable auxiliar arranca con el primer elemento
for i in range(1, len(lista)):   # arrancamos desde el segundo
    actual = lista[i]
    # comparamos actual con anterior
    anterior = actual             # al final, guardamos el actual para la próxima vuelta
```

La última línea (`anterior = actual`) es el mismo gesto que el swap: "me guardo este valor antes de pasar al siguiente". Sirve para detectar cambios, subidas, bajadas, repeticiones...

````python
# Ejemplo: contar cuántas veces la temperatura subió respecto al día anterior
temps = [18, 22, 20, 25, 31, 29, 32]

subidas = 0
anterior = temps[0]
for i in range(1, len(temps)):
    if temps[i] > anterior:
        subidas += 1
    anterior = temps[i]

print('Días que subió:', subidas)  # 4
````

## Patrón 5: Secuencia más larga

Este patrón combina los anteriores. Necesitamos rastrear **dos cosas al mismo tiempo**:
- la racha **actual** (que puede romperse)
- la mejor racha **hasta ahora** (que solo crece)

```python
racha_actual = 1
mejor_racha  = 1

for i in range(1, len(lista)):
    if condicion(lista[i], lista[i-1]):
        racha_actual += 1              # la racha sigue
    else:
        racha_actual = 1               # la racha se rompe: reset

    if racha_actual > mejor_racha:
        mejor_racha = racha_actual     # actualizamos el récord
```

La clave está en el **reset**: cuando la condición no se cumple, `racha_actual` vuelve a 1 (no a 0, porque el elemento actual solo forma una racha de longitud 1 consigo mismo).

````python
# Ejemplo: secuencia más larga de números consecutivos crecientes
# (cada elemento mayor que el anterior)
nums = [1, 3, 5, 2, 4, 6, 8, 1, 9]
#             ↑↑↑ se rompe       ↑ se rompe
#       racha=3         racha=4

racha_actual = 1
mejor_racha  = 1

for i in range(1, len(nums)):
    if nums[i] > nums[i-1]:
        racha_actual += 1
    else:
        racha_actual = 1

    if racha_actual > mejor_racha:
        mejor_racha = racha_actual

print('Mejor racha creciente:', mejor_racha)  # 4
````

````python
# Misma lógica pero con traza: imprimimos el estado en cada paso
# Útil para entender qué está pasando
nums = [1, 3, 5, 2, 4, 6, 8, 1, 9]

racha_actual = 1
mejor_racha  = 1

for i in range(1, len(nums)):
    if nums[i] > nums[i-1]:
        racha_actual += 1
    else:
        racha_actual = 1

    if racha_actual > mejor_racha:
        mejor_racha = racha_actual

    print(f'i={i}  actual={nums[i]}  racha={racha_actual}  mejor={mejor_racha}')
````

### Ejercicio F
Encontrá la secuencia más larga de letras **en orden alfabético** consecutivo en una cadena.

Ejemplo: `'abcxyzde'` → `'xyz'` (longitud 3)

Pista: dos letras están en orden alfabético consecutivo si `ord(b) == ord(a) + 1`

````python
cadena = 'abcxyzde'

racha_actual = 1
mejor_racha  = 1
inicio_mejor = 0   # posición donde empieza la mejor racha
inicio_actual = 0

for i in range(1, len(cadena)):
    if ord(cadena[i]) == ord(cadena[i-1]) + 1:
        racha_actual += 1
    else:
        racha_actual  = 1
        inicio_actual = i

    if racha_actual > mejor_racha:
        mejor_racha  = racha_actual
        inicio_mejor = inicio_actual

mejor_secuencia = cadena[inicio_mejor : inicio_mejor + mejor_racha]
print(f'Mejor secuencia: {mejor_secuencia!r} (longitud {mejor_racha})')
# 'xyz' (longitud 3)

# Probá con otras cadenas:
# 'mnopqabc' → ?
# 'zyx'      → ?
````

### Ejercicio G — el ejercicio del jueves ⭐

Encontrá la **subsecuencia más larga de números estrictamente crecientes** en una lista, y reportá:
- su longitud
- en qué posición empieza
- los elementos que la forman

Ejemplo: `[3, 1, 2, 5, 4, 7, 8, 9, 2]`
- La racha más larga es `4, 7, 8, 9` → longitud 4, empieza en posición 4

````python
nums = [3, 1, 2, 5, 4, 7, 8, 9, 2]

racha_actual  = 1
mejor_racha   = 1
inicio_actual = 0
inicio_mejor  = 0

for i in range(1, len(nums)):
    if nums[i] > nums[i-1]:
        racha_actual += 1
    else:
        racha_actual  = 1
        inicio_actual = i

    if racha_actual > mejor_racha:
        mejor_racha  = racha_actual
        inicio_mejor = inicio_actual

mejor_subsec = nums[inicio_mejor : inicio_mejor + mejor_racha]
print(f'Longitud: {mejor_racha}')
print(f'Empieza en posición: {inicio_mejor}')
print(f'Elementos: {mejor_subsec}')
# Longitud: 4
# Empieza en posición: 4
# Elementos: [4, 7, 8, 9]
````

## Resumen: cómo elegir el patrón

Cuando te enfrentás a un problema con listas, preguntate:

```
¿Necesito intercambiar dos valores?            →  Swap (variable auxiliar)
¿Necesito calcular un total o construir algo?  →  Acumulador
¿Necesito saber si algo ocurrió?               →  Bandera
¿Necesito el mejor/peor valor?                 →  Máximo / mínimo
¿Necesito comparar cada elemento con el anterior?  →  Variable auxiliar
¿Necesito la racha más larga de algo?          →  Secuencia más larga
```

Muchos problemas combinan más de uno.

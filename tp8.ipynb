## ¿Qué imprime?

Deducir el comportamiento antes de correr el bloque.

## 1

```python
# ¿Qué imprime?
def f(n):
    if n == 0:
        return 0
    return n + f(n - 1)

print(f(5))
```

## 2

```python
# ¿Qué imprime?
def g(lista):
    if len(lista) == 0:
        return 0
    return lista[0] + g(lista[1:])

print(g([3, 1, 4, 1, 5]))
```

## 3

```python
# ¿Qué imprime?
def h(n):
    if n == 0:
        return
    print(n)
    h(n - 1)

h(4)
```

## 4

```python
# ¿Qué imprime? ¿En qué se diferencia del ejercicio 3?
def h2(n):
    if n == 0:
        return
    h2(n - 1)
    print(n)

h2(4)
```

## 5

```python
# ¿Qué imprime?
def misterio(lista):
    if len(lista) == 0:
        return []
    return misterio(lista[1:]) + [lista[0]]

print(misterio([1, 2, 3, 4, 5]))
```

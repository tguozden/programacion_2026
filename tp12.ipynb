## Comprensión de listas

La comprensión de listas es una forma compacta de construir una lista a partir de un iterable:

```python
# Con for clásico:
cuadrados = []
for x in range(5):
    cuadrados.append(x ** 2)

# Con comprensión:
cuadrados = [x ** 2 for x in range(5)]
# [0, 1, 4, 9, 16]
```

La estructura general es:

```python
[expresion for variable in iterable]
```

Se puede agregar un filtro con `if`:

```python
[expresion for variable in iterable if condicion]
```

Ejemplo:

```python
[x for x in range(10) if x % 2 == 0]
# [0, 2, 4, 6, 8]
```

---
## Comprensión de conjuntos y diccionarios

La misma idea funciona para conjuntos y diccionarios cambiando los corchetes:

```python
# Conjunto — sin duplicados, sin orden:
{x ** 2 for x in range(5)}
# {0, 1, 4, 9, 16}

# Diccionario — clave: valor:
{x: x ** 2 for x in range(5)}
# {0: 0, 1: 1, 2: 4, 3: 9, 4: 16}
```

También aceptan filtro con `if`:

```python
{x: x ** 2 for x in range(10) if x % 2 == 0}
# {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}
```

## Comprensiones anidadas

Se pueden tener varios `for` dentro de una comprensión. El orden es el mismo que escribirías los bucles anidados:

```python
# Con for clásico:
pares = []
for x in [1, 2, 3]:
    for y in ['a', 'b']:
        pares.append((x, y))

# Con comprensión:
pares = [(x, y) for x in [1, 2, 3] for y in ['a', 'b']]
# [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b'), (3, 'a'), (3, 'b')]
```

**Criterio:** una comprensión es buena cuando se lee de corrido como una oración. Si necesitás comentarla para entenderla, probablemente sea mejor un `for` clásico.

## 8
### Condicional en la expresión

La comprensión también admite un `if/else` dentro de la *expresión* (no del filtro). Son dos cosas distintas:

```python
# Filtro: descarta elementos
[x for x in lista if condicion]

# Condicional en expresión: transforma todos los elementos
[a if condicion else b for x in lista]
```

Antes de ejecutar, predecí qué imprime cada línea.

````python
nums = [-3, -1, 0, 2, 4]

print([x if x >= 0 else -x for x in nums])
print(['par' if x % 2 == 0 else 'impar' for x in range(6)])
print([x ** 2 if x > 0 else 0 for x in nums])
````

```python
def miswap(lista, i, j):
    aux = lista[i]
    lista[i] = lista[j]
    lista[j] = aux
print(lista)
miswap(lista, 0, 1)
print(lista)
    
```

```python
lista = [2,7,4,-3,2,0,1]
print(lista)
def seleccion(lista):
    n = len(lista)
    for i in range(n-1):
        imin = i
        for j in range(i,n):  #bucle para buscar el mínimo
            if lista[j] < lista[imin]:
                imin = j
        if i != imin:
            miswap(lista, i, imin)
        print(lista)
seleccion(lista)
```

```python
# ordenamiento por insercion
lista = [2,7,4,-3,2,0,1]

def insercion_aux(lista):
    aux = []
    for elemento in lista:
        aux.append(elemento)
        i = len(aux) - 1   #posición del que acabo de agregar
        while i > 0 and aux[i-1] > elemento:
            i = i - 1
        miswap(aux, i, len(aux)-1)
        print(aux)
    for i in range(len(lista)):
        lista[i] = aux[i]
insercion_aux(lista)
print(lista)
```

```python
# ordenamiento por insercion
lista = [2,7,4,-3,2,0,1]

def insercion(lista):
    n = len(lista)
    for j in range(1,n):

        i = j   #posición del que acabo de agregar
        while i > 0 and lista[i-1] > lista[j]:
            i = i - 1
        miswap(lista, i, j)
        print(lista)
insercion(lista)
print(lista)
```

```python
lista = [2,7,4,-3,2,0,1]
# vamos a burbujear
print(lista)
for i in range(len(lista) - 1):
    print(f"i= {i}", end=' ')
    if lista[i] > lista[i+1]:
        miswap(lista, i, i + 1)
    print(lista)
    
```

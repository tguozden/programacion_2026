# 2
Construya el código que busca los factores de
un número ingresado por el usuario y los imprime
en pantalla. 

Pista 1: recorra los números desde
1 hasta `n` inclusive. Luego averigüe si dicho número es divisor de `n`,
si es así imprímalo.
    
Pista 2: si un número `x` es divisor de `n` entonces el resto de la división `n/x` es cero.

```python
n = int(input("ingrese un número"))

#  su código
```



````python
n = 10
for i in range(1, n + 1):
    print(f"el resto de {n} dividido {i} es {10%i}")
````

````python
n = 10
for i in range(1, n + 1):
    if n%i == 0:
        print(i, end=' - ')
````

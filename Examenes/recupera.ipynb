### Orden por promedio

Escribí una función `orden_promedio(notas)` que devuelva una lista de enteros indicando el orden de los alumnos **de mayor a menor promedio**.

Por ejemplo, si el alumno con mayor promedio es el de la posición `4`, el primer elemento de la lista será `4`.

Resultado esperado para la cadena de arriba:
```python
[4, 2, 1, 5, 0, 3]
```

Pista: podés usar el diccionario `proms` del ejercicio anterior.

````python
notas = [(7.5, 8.0), (6.0, 9.5), (8.5, 7.0), (5.0, 6.5), (9.0, 8.0), (7.0, 7.5)]

prom = {tupla:sum(tupla)/2 for tupla in notas}
nota = prom.values()
print(nota)
````

````python
def orden_promedio(nota):   #prom.values()
    lista = [(i,j) for i,j in enumerate(nota)]
    n = len(lista)
    for i in range(n-1):
        for j in range(n-1-i):
            if lista[j][1]>lista[j+1][1]:
                aux=lista[j]
                lista[j]=lista[j+1]
                lista[j+1]=aux
            # print(lista)
    return [i[0] for i in lista]

print(nota)
orden_promedio(nota)
````

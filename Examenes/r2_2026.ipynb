### Asimov (6 puntos)
Tomemos nuevamente el texto `asimov.txt`. Usando la lista de palabras del texto en minúscula que generamos, imprima cuántas palabras hay en total y cuántas de ellas tiene caracteres especiales, esto es que no están entre  `ord('a')=97` y `ord(z)=122`.

````python
#código que generamos el martes:
with open("asimov.txt", "r", encoding="utf-8") as file:
    texto = file.read()
lista = texto.split()
lista = [i.lower() for i in lista]

def limpia(palabra:str)->str:
    ''' esta función quita los caracteres
        que no son letras, como ? - ! ¿ comillas, comas, espacios, etc.
    '''
    letras = [i.lower() for i in palabra if i.isalpha()]
    return ''.join(letras)

lista = [limpia(i) for i in lista]
````
